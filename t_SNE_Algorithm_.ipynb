## **Simple** Linear Regression *


Simple Linear Regression is used to predict a continuous value using one independent variable.

Formula --> y=mx+c

Where

x → Independent Variable, y → Dependent Variable, m → Slope, c → Intercept


[ ]


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Dataset
df = pd.DataFrame({
    "Experience": [1, 2, 3, 4, 5],
    "Salary": [3, 4, 5, 6, 7]
})

# Independent Variable
X = df[["Experience"]]

# Dependent Variable
y = df["Salary"]

# Create and Train Model
model = LinearRegression()
model.fit(X, y)

# Predict values for the regression line
y_pred = model.predict(X)

# Predict salary for 6 years of experience
prediction = model.predict([[6]])

# Visualization
plt.figure(figsize=(6,4))
plt.scatter(df["Experience"], df["Salary"], color="blue", s=80, label="Actual Data")
plt.plot(df["Experience"], y_pred, color="red", linewidth=2, label="Regression Line")
plt.scatter(6, prediction[0], color="green", s=100, marker="*", label="Prediction (6, 8)")

plt.title("Simple Linear Regression")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.legend()
plt.grid(True)

plt.show()




```

# **Multiple** Linear Regression
What is it?

Multiple Linear Regression predicts the output using more than one input feature.

Formula Y=b0+b1X1+b2X2+...+bnXn​

Example

Predict House Price

Inputs

Area, Bedrooms, Age,

Output

House Price

```python
import matplotlib.pyplot as plt

# Predict prices for all houses
y_pred = model.predict(X)

# Plot Actual vs Predicted
plt.scatter(y, y_pred, color="blue", s=80)

# Ideal prediction line
plt.plot([y.min(), y.max()],
         [y.min(), y.max()],
         color="red")

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Multiple Linear Regression")
plt.grid(True)

plt.show()
```

# **Logistic** Regression (Classification)
What is Logistic Regression?

Logistic Regression is a Supervised Machine Learning Classification algorithm used to predict categorical outcomes, especially Binary Classification.

Although its name contains "Regression", it is used for Classification, not Regression.

Why is it called Logistic Regression?
Unlike Linear Regression, which predicts continuous values, Logistic Regression predicts the probability of belonging to a class.

It uses the Sigmoid Function to convert any value into a probability between 0 and 1.


[ ]


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

# Dataset
df = pd.DataFrame({
    "Hours": [1, 2, 3, 4, 5],
    "Pass":  [0, 0, 0, 1, 1]
})

# Independent Variable
X = df[["Hours"]]

# Dependent Variable
y = df["Pass"]

# Create Logistic Regression Model
model = LogisticRegression()

# Train the Model
model.fit(X, y)

# Predict for 3.5 hours
prediction = model.predict([[3.5]])
probability = model.predict_proba([[3.5]])

print("Prediction (0 = Fail, 1 = Pass):", prediction[0])
print("Probability:", probability[0])

# Visualization
plt.figure(figsize=(6,4))

# Plot Actual Data
plt.scatter(df["Hours"], df["Pass"], color="blue", s=80, label="Actual Data")

# Logistic Curve
x_values = np.linspace(1, 6, 100).reshape(-1, 1)
y_prob = model.predict_proba(x_values)[:, 1]

plt.plot(x_values, y_prob, color="red", linewidth=2, label="Logistic Curve")

# Plot Prediction
plt.scatter(3.5, probability[0][1], color="green", s=150,
            marker="*", label="Prediction")

plt.xlabel("Hours Studied")
plt.ylabel("Probability of Passing")
plt.title("Logistic Regression")
plt.legend()
plt.grid(True)

plt.show()
```

# **K**-Nearest Neighbors (KNN) in Python
Definition

K-Nearest Neighbors (KNN) is a supervised machine learning algorithm used for classification and regression. It predicts the class of a new data point based on the majority class of its K nearest neighbors.*italicised text*

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier

# Dataset
df = pd.DataFrame({
    "Hours": [1, 2, 3, 4, 5],
    "Pass": [0, 0, 1, 1, 1]
})

# Independent Variable
X = df[["Hours"]]

# Dependent Variable
y = df["Pass"]

# Create KNN Model
model = KNeighborsClassifier(n_neighbors=3)

# Train Model
model.fit(X, y)

# Prediction
prediction = model.predict([[3.5]])

print("Prediction (0 = Fail, 1 = Pass):", prediction[0])

# Visualization
plt.figure(figsize=(6,4))

# Plot Actual Data
plt.scatter(df["Hours"], df["Pass"], color="blue", s=80, label="Actual Data")

# Plot Predicted Point
plt.scatter(3.5, prediction[0], color="red", s=150, marker="*", label="Prediction")

plt.xlabel("Hours Studied")
plt.ylabel("Pass (0 = Fail, 1 = Pass)")
plt.title("K-Nearest Neighbors (KNN)")
plt.xticks([1, 2, 3, 3.5, 4, 5])
plt.yticks([0, 1], ["Fail", "Pass"])
plt.legend()
plt.grid(True)

plt.show()


```

# **Polynomial** Regression

Polynomial Regression is a machine learning algorithm that models the relationship between an independent variable and a dependent variable using a polynomial equation. It is useful when the relationship between variables is non-linear.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Dataset
df = pd.DataFrame({
    "X": [1, 2, 3, 4, 5],
    "Y": [1, 4, 9, 16, 25]
})

# Independent Variable
X = df[["X"]]

# Dependent Variable
y = df["Y"]

# Create Polynomial Features (Degree = 2)
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# Create and Train Model
model = LinearRegression()
model.fit(X_poly, y)

# Predict for X = 6
prediction = model.predict(poly.transform([[6]]))

print("Predicted Value:", prediction[0])

# Create Smooth Curve
x_values = np.linspace(1, 6, 100).reshape(-1, 1)
x_values_poly = poly.transform(x_values)
y_values = model.predict(x_values_poly)

# Visualization
plt.figure(figsize=(7,5))

# Actual Data
plt.scatter(df["X"], df["Y"], color="blue", s=80, label="Actual Data")

# Polynomial Curve
plt.plot(x_values, y_values, color="red", linewidth=2, label="Polynomial Curve")

# Predicted Point
plt.scatter(6, prediction[0], color="green", s=150, marker="*", label="Prediction")

# Labels
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Polynomial Regression")
plt.legend()
plt.grid(True)

# Show Graph
plt.show()
```

# **Naive Bayes Algorithm **

Naive Bayes is a supervised machine learning classification algorithm based on Bayes' Theorem. It is commonly used for spam detection, text classification, and sentiment analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.naive_bayes import GaussianNB

# Dataset
df = pd.DataFrame({
    "Hours": [1, 2, 3, 4, 5, 6],
    "Pass":  [0, 0, 0, 1, 1, 1]
})

# Independent Variable
X = df[["Hours"]]

# Dependent Variable
y = df["Pass"]

# Create Naive Bayes Model
model = GaussianNB()

# Train Model
model.fit(X, y)

# Prediction
prediction = model.predict([[3.5]])

print("Prediction (0 = Fail, 1 = Pass):", prediction[0])

# Visualization
plt.figure(figsize=(6,4))

# Actual Data
plt.scatter(df["Hours"], df["Pass"], color="blue", s=80, label="Actual Data")

# Predicted Point
plt.scatter(3.5, prediction[0], color="red", marker="*", s=150, label="Prediction")

plt.xlabel("Hours Studied")
plt.ylabel("Pass (0 = Fail, 1 = Pass)")
plt.title("Naive Bayes Classification")
plt.xticks([1, 2, 3, 3.5, 4, 5, 6])
plt.yticks([0, 1], ["Fail", "Pass"])
plt.legend()
plt.grid(True)

plt.show()
```

# **Decision Tree Regression **


Decision Tree Regression is a supervised machine learning algorithm used to predict continuous numerical values such as salary, house prices, sales, temperature, and marks. It builds a tree-like model by repeatedly splitting the dataset into smaller groups based on feature values. Each final group (leaf node) contains the predicted numerical value.

Unlike Linear Regression, which fits a straight line, Decision Tree Regression divides the data into regions and predicts a constant value for each region, resulting in a step-shaped prediction curve.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.tree import DecisionTreeRegressor

# Dataset
df = pd.DataFrame({
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [10, 20, 30, 40, 50]
})

# Independent Variable
X = df[["Hours"]]

# Dependent Variable
y = df["Marks"]

# Create Decision Tree Regression Model
model = DecisionTreeRegressor(random_state=0)

# Train Model
model.fit(X, y)

# Prediction
prediction = model.predict([[3.5]])

print("Predicted Marks:", prediction[0])

# Visualization
x_grid = np.arange(min(df["Hours"]), max(df["Hours"]) + 0.01, 0.01)
x_grid = x_grid.reshape((len(x_grid), 1))

plt.figure(figsize=(6,4))

# Actual Data
plt.scatter(df["Hours"], df["Marks"], color="blue", s=80, label="Actual Data")

# Decision Tree Regression Line
plt.plot(x_grid, model.predict(x_grid), color="red", label="Decision Tree Regression")

# Prediction Point
plt.scatter(3.5, prediction[0], color="green", marker="*", s=150, label="Prediction")

plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Decision Tree Regression")
plt.legend()
plt.grid(True)

plt.show()
```
