## Matplotlib is a popular Python library used for data visualization. It helps create graphs, charts, and plots to represent data in a clear and understandable way.

Definition

Matplotlib is an open-source Python library used to create static, animated, and interactive visualizations such as line charts, bar charts, pie charts, histograms, and scatter plots.

Features of Matplotlib
Creates different types of graphs and charts.
Helps visualize and analyze data easily.
Supports customization of titles, labels, colors, and styles.
Works well with NumPy and Pandas.
# Widely used in data analysis, machine learning, and scientific computing  **Common Graphs in Matplotlib**
Line Plot,
Bar Chart,
Pie Chart,
Histogram,
Scatter Plot,
Box Plot,

```python
import matplotlib.pyplot as plt

# Create dataset
students = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']
marks = [78, 85, 92, 67, 88, 96, 72, 81, 90, 76]

# Display students and marks
print("Student Marks")
for i in range(len(students)):
    print(students[i], ":", marks[i])

# Find the highest scorer
highest_mark = max(marks)
highest_student = students[marks.index(highest_mark)]

print("\nHighest Scorer:")
print(highest_student, "-", highest_mark)

# Plot in ascending order
plt.figure(figsize=(8,5))
plt.bar(students, marks)
plt.title("Student Marks Analysis (Ascending Order)")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()

# Display marks in descending order
students_desc = [x for _, x in sorted(zip(marks, students), reverse=True)]
marks_desc = sorted(marks, reverse=True)

print("\nMarks in Descending Order:")
for i in range(len(students_desc)):
    print(students_desc[i], ":", marks_desc[i])

# Plot in descending order
plt.figure(figsize=(8,5))
plt.bar(students_desc, marks_desc)
plt.title("Student Marks Analysis (Descending Order)")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```
