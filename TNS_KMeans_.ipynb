### K-Means Clustering

# Definition:
K-Means is an unsupervised machine learning algorithm used to divide a dataset into K clusters based on the similarity of data points. It groups similar data together and separates dissimilar data.

# Steps of the K-Means Algorithm
1.Choose the number of clusters (K).

2.Randomly select K centroids.

3.Assign each data point to the nearest centroid.

4.Calculate new centroids by finding the mean of all points in each cluster.

5.Repeat steps 3 and 4 until the centroids no longer change or the maximum number of iterations is reached.

 Elbow Method

# Definition:
The Elbow Method is a technique used to determine the optimal number of clusters (K) for the K-Means clustering algorithm. It identifies the value of K at which adding more clusters provides only a small improvement.

# How It Works
# Choose a range of values for K (e.g., 1 to 10).
Run the K-Means algorithm for each value of K.
Calculate the Within-Cluster Sum of Squares (WCSS) (also called inertia) for each value of K.
Plot K on the x-axis and WCSS on the y-axis.
Select the value of K at the "elbow" point, where the decrease in WCSS begins to slow down.
# Why is it Called the Elbow Method?
The graph usually looks like a bent arm. The point where the curve bends sharply (forming an "elbow") is considered the best number of clusters

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Sample dataset
data = {
    "Age": [25, 30, 35, 40, 45, 50, 28, 32, 38, 42],
    "Income": [25000, 30000, 45000, 50000, 60000, 70000, 28000, 35000, 48000, 55000]
}

# Create DataFrame
df = pd.DataFrame(data)

# Select features
X = df[["Age", "Income"]]

# Calculate WCSS for different values of K
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

# Plot the Elbow Method graph
plt.figure(figsize=(6, 5))
plt.plot(range(1, 11), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS (Inertia)")
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()
```
