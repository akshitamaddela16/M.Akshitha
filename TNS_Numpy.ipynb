# **NumPy** (Numerical Python)
**Numpy is a fundamental Python library used for numerical computing. It provides support for large, multi-dimensional arrays and matrices, along with a collection of mathematical functions to operate on them efficiently.

# key feature of numpy
N-dimensional arrays (ndarray) for storing and manipulating data.
Fast mathematical operations such as addition, subtraction, multiplication, and division.
Statistical functions like mean, median, standard deviation, and variance.
Linear algebra operations, including matrix multiplication and solving equations.
Random number generation for simulations and data analysis.
Efficient memory usage and high performance compared to standard Python lists.

# create a numpy array containing numbers from 1 to 20 using inbuilt function

```python
import numpy as np

# Create a NumPy array from 1 to 20
arr = np.arange(1, 21)

print(arr)
```

# create an a array of all zero's with 10 elements and convert it to matrix

```python
import numpy as np

# Create an array of 10 zeros
arr = np.zeros(10)

# Convert the array into a 2x5 matrix
matrix = arr.reshape(2, 5)

print("Array:")
print(arr)

print("\nMatrix:")
print(matrix)
```

#create an a array containing even numbers from 2 to 20

```python
import numpy as np

# Create an array of even numbers from 2 to 20
arr = np.arange(2, 21, 2)

print(arr)
```

# create an array and find number of dimensions shape, size,data type and multiple every element by 2

```python
import numpy as np

# Create a NumPy array
arr = np.array([10, 20, 30, 40, 50])

# Number of dimensions
print("Number of Dimensions:", arr.ndim)

# Shape of the array
print("Shape:", arr.shape)

# Size of the array
print("Size:", arr.size)

# Data type of the array
print("Data Type:", arr.dtype)

# Multiply every element by 2
result = arr * 2

print("Original Array:", arr)
print("Array after multiplying by 2:", result)
```

# create id array and find some mean,max,min value

```python
import numpy as np

# Create a 1D array
arr = np.array([10, 20, 30, 40, 50])

# Find sum
print("Sum:", np.sum(arr))

# Find mean
print("Mean:", np.mean(arr))

# Find maximum value
print("Maximum:", np.max(arr))

# Find minimum value
print("Minimum:", np.min(arr))
```
